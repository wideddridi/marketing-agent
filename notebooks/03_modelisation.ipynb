```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
# Chargement des données chronologiques
train_c = pd.read_csv("../data/clean/chronologique/train.csv")
valid_c = pd.read_csv("../data/clean/chronologique/val.csv")
test_c = pd.read_csv("../data/clean/chronologique/test.csv")
```

```python
# Séparation des X et Y, et élimination des colonnes inutiles
colonnes_exclues = ["y", "y_num", "duration"]

# Supprimer les colonnes inutiles pour les features de puis les datasets train, valid et test
X_train_c = train_c.drop(columns=colonnes_exclues)
y_train_c = train_c["y_num"]

X_valid_c = valid_c.drop(columns=colonnes_exclues)
y_valid_c = valid_c["y_num"]

X_test_c = test_c.drop(columns=colonnes_exclues)
y_test_c = test_c["y_num"]

print(X_train_c.shape, y_train_c.shape)  # (28823, 21) (28823,)
```

```python
# Lister les colonnes selon le modele
colonnes_cat = X_train_c.select_dtypes(exclude="number").columns.tolist()

colonnes_num_lr = ["age", "campaign_plafonnee", "previous", "pdays_jours",
                   "deja_contacte", "euribor3m"]

colonnes_num_xgb = ["age", "campaign", "previous", "pdays_jours", "deja_contacte",
                    "emp.var.rate", "cons.price.idx", "cons.conf.idx",
                    "euribor3m", "nr.employed"]
```

```python
def construire_preprocesseur(colonnes_num, colonnes_cat, normaliser=True):
    etape_num = StandardScaler() if normaliser else "passthrough"
    return ColumnTransformer(
        transformers=[
            ("num", etape_num, colonnes_num),
            ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), colonnes_cat),
        ]
    )
```
